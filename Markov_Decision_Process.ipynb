```python
import copy
import timeit
import numpy as np
import matplotlib.pyplot as pl
%matplotlib inline
from ipywidgets import interactive
import ipywidgets as widgets
```

Create the following grid world.

**States:** A 10x10 grid

**Actions:** Up, Down, Left, Right

**Tranistion probabilities:**
* 0.7 in the direction of action
* 0.1 in the three other directions
* The robot bounces back to the same state near edges

**Rewards:**
* (7,8) has a reward +10
* (2,7) has a reward +3
* (4,3) has a reward -5
* (7,3) has a reward -10
* No reward in other states

This example is based on Decision Making Under Uncertainty by M.J. Kochenderfer.

```python
#Let's define MDP paras
def createGrid10World():
    def xy2s(y, x):
        x = max(x, 0)
        y = max(y, 0)
        x = min(x, 9)
        y = min(y, 9)
        out = np.ravel_multi_index(np.array([x,y]), (10,10))
        return out

    def s2xy(s):
        x, y = np.unravel_index(s, (10,10))
        return y, x

    def gridPlot(ax, im, title='', cmap='Blues'):
        pl.imshow(im, interpolation='none', cmap=cmap, origin='lower')
        pl.colorbar()
        ax.set_xticks(np.arange(0, 10, 1));
        ax.set_yticks(np.arange(0, 10, 1));
        ax.set_xticklabels(np.arange(0, 10, 1));
        ax.set_yticklabels(np.arange(0, 10, 1));
        ax.set_xticks(np.arange(-.5, 10, 1), minor=True);
        ax.set_yticks(np.arange(-.5, 10, 1), minor=True);
        ax.grid(which='minor', color='w', linestyle='-', linewidth=1)
        pl.title(title);
        return

    A = ['left', 'right', 'up', 'down']
    S = np.arange(100)
    T = np.zeros((len(S), len(A), len(S)))
    R = np.zeros((len(S), len(A)))
    for s in S:
        x, y = s2xy(s)
        if x == 2 and y == 7:
            R[s, :] = 3
        elif x == 7 and y == 8:
            R[s, :] = 10
        else:
            if x == 7 and y == 3:
                R[s, :] = -10
            elif x == 4 and y == 3:
                R[s, :] = -5
            elif x == 0:
                if y == 0 or y == 9:
                    R[s, :] = -0.2
                else:
                    R[s, :] = -0.1
                R[s, 0] = -0.7
            elif x == 9:
                if y == 0 or y == 9:
                    R[s, :] = -0.2
                else:
                    R[s, :] = -0.1
                R[s, 1] = -0.7
            elif y == 0:
                if x == 0 or x == 9:
                    R[s, :] = -0.2
                else:
                    R[s, :] = -0.1
                R[s, 3] = -0.7
            elif y == 9:
                if x == 0 or x == 9:
                    R[s, :] = -0.2
                else:
                    R[s, :] = -0.1
                R[s, 2] = -0.7

            for a in A:
                if a == 'left':
                    T[s, 0, xy2s(x - 1, y)] += 0.7
                    T[s, 0, xy2s(x + 1, y)] += 0.1
                    T[s, 0, xy2s(x, y - 1)] += 0.1
                    T[s, 0, xy2s(x, y + 1)] += 0.1
                elif a == 'right':
                    T[s, 1, xy2s(x + 1, y)] += 0.7
                    T[s, 1, xy2s(x - 1, y)] += 0.1
                    T[s, 1, xy2s(x, y - 1)] += 0.1
                    T[s, 1, xy2s(x, y + 1)] += 0.1
                elif a == 'up':
                    T[s, 2, xy2s(x, y + 1)] += 0.7
                    T[s, 2, xy2s(x, y - 1)] += 0.1
                    T[s, 2, xy2s(x - 1, y)] += 0.1
                    T[s, 2, xy2s(x + 1, y)] += 0.1
                elif a == 'down':
                    T[s, 3, xy2s(x, y - 1)] += 0.7
                    T[s, 3, xy2s(x, y + 1)] += 0.1
                    T[s, 3, xy2s(x - 1, y)] += 0.1
                    T[s, 3, xy2s(x + 1, y)] += 0.1

    for a, c_x, c_y in [(0,0,0), (0,0,9),(1,9,0),(1,9,9),(2,0,9),(2,9,9),(3,0,0),(3,9,0)]:
        R[xy2s(c_x,c_y),a] = -0.8

    discount = 0.9

    nextStates = {}
    for si in range(len(S)):
        for ai in range(len(A)):
            nextStates[(si,ai)] = np.where((T[si, ai, :] != 0) == True)[0]

    return {'S':S, 'A':A, 'T':T, 'R':R, 'discount':discount, 'nextStates':nextStates, 'gridPlot':gridPlot, 'xy2s':xy2s, 's2xy':s2xy}
```

```python
class MDP():
    def __init__(self):
        pass

    def inbuilt_init(self, mdp_dict):
        self.S = mdp_dict['S']
        self.A = mdp_dict['A']
        self.T = mdp_dict['T']
        self.R = mdp_dict['R']
        self.discount = mdp_dict['discount']
        self.nextStates = mdp_dict['nextStates']
        self.gridPlot = mdp_dict['gridPlot']
        self.xy2s = mdp_dict['xy2s']
        self.s2xy = mdp_dict['s2xy']
```

```python
#Define the MDP
mdp = MDP()
mdp.inbuilt_init(mdp_dict=createGrid10World())

#Plot states
pl.figure(figsize=(3,3))
mdp.gridPlot(ax=pl.gca(), im=mdp.S.reshape((10,10)), title='States', cmap='Greys')

#Plot rewards
pl.figure(figsize=(15,3))
pl.suptitle('Rewards', fontsize=15)
for a in range(4):
    pl.subplot(1,4,a+1)
    mdp.gridPlot(ax=pl.gca(), im=mdp.R[:,a].reshape((10,10)), title='a='+mdp.A[a], cmap='jet')
pl.show()

#Plot rewards
pl.figure(figsize=(15,3))
pl.suptitle('Rewards - clipped larger +ve values for visualizing the edges', fontsize=15)
for a in range(4):
    pl.subplot(1,4,a+1)
    mdp.gridPlot(ax=pl.gca(), im=np.clip(mdp.R[:,a].reshape((10,10)), -1, -0.5), title='a='+mdp.A[a], cmap='jet')
pl.show()

#Plot rewards
s0_x, s0_y = 3, 5
s0 = mdp.xy2s(s0_x, s0_y)
pl.figure(figsize=(15,3))
pl.suptitle('Transition probabilities T(s1|s0=({},{}),a)'.format(s0_x, s0_y), fontsize=15)
for a in range(4):
    pl.subplot(1,4,a+1)
    mdp.gridPlot(ax=pl.gca(), im=mdp.T[s0,a,:].reshape((10,10)), title='a='+mdp.A[a], cmap='Blues')
pl.show()
```

```python
import time
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(4050)

# The supplied notebook leaves transitions empty at reward states.
# Give those states the same 0.7/0.1 movement rule as other states.
for s in mdp.S:
    for a in range(len(mdp.A)):
        if mdp.T[s, a, :].sum() == 0:
            x, y = mdp.s2xy(s)
            moves = {
                0: [(x-1, y), (x+1, y), (x, y-1), (x, y+1)],
                1: [(x+1, y), (x-1, y), (x, y-1), (x, y+1)],
                2: [(x, y+1), (x, y-1), (x-1, y), (x+1, y)],
                3: [(x, y-1), (x, y+1), (x-1, y), (x+1, y)]
            }
            for probability, (nx, ny) in zip(
                [0.7, 0.1, 0.1, 0.1], moves[a]
            ):
                mdp.T[s, a, mdp.xy2s(nx, ny)] += probability

# ---------- Model based: Value Iteration ----------
start = time.perf_counter()
U = np.zeros(len(mdp.S))
vi_changes = []

for sweep in range(500):
    action_values = (
        mdp.R
        + mdp.discount * np.einsum("sak,k->sa", mdp.T, U)
    )
    U_new = np.max(action_values, axis=1)

    change = np.max(np.abs(U_new - U))
    vi_changes.append(change)
    U = U_new

    if change < 0.001:
        break

vi_time = time.perf_counter() - start

# ---------- Model free: Q-learning ----------
# Q-learning uses sampled rewards and next states to update Q.
# It does not use the transition matrix in its update equation.
rng = np.random.default_rng(4050)
Q = np.zeros((len(mdp.S), len(mdp.A)))
q_errors = []

start = time.perf_counter()

for episode in range(5000):
    state = int(rng.integers(len(mdp.S)))

    for step in range(100):
        if rng.random() < 0.25:
            action = int(rng.integers(len(mdp.A)))
        else:
            action = int(np.argmax(Q[state]))

        next_state = int(
            rng.choice(mdp.S, p=mdp.T[state, action, :])
        )
        reward = mdp.R[state, action]

        Q[state, action] += 0.12 * (
            reward
            + mdp.discount * np.max(Q[next_state])
            - Q[state, action]
        )
        state = next_state

    # Diagnostic only: compare learned values with Value Iteration.
    error = np.mean(np.abs(np.max(Q, axis=1) - U))
    q_errors.append(error)

q_time = time.perf_counter() - start

print(f"Value Iteration: {len(vi_changes)} sweeps")
print(f"Execution time: {vi_time:.3f} seconds")
print(f"Final maximum value change: {vi_changes[-1]:.5f}")
print()
print("Q-learning: 5000 episodes")
print(f"Execution time: {q_time:.3f} seconds")
print(f"Final mean value error: {q_errors[-1]:.3f}")

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(vi_changes)
axes[0].axhline(0.001, color="red", linestyle="--")
axes[0].set_yscale("log")
axes[0].set_title("Value Iteration convergence")
axes[0].set_xlabel("Sweep")
axes[0].set_ylabel("Maximum change in U")

axes[1].plot(q_errors)
axes[1].set_title("Q-learning convergence")
axes[1].set_xlabel("Episode")
axes[1].set_ylabel("Mean value error against Value Iteration")

plt.tight_layout()
plt.show()
```

### 1. Policy evaluation

Computing the utility, U.

$U^\pi_k(s) = R(s, \pi(s)) + \gamma \sum_{s'} T(s' \mid s, \pi(s))U^\pi_{k-1}(s')$

```python
def iterativePolicyEvaluation(mdp, policy, numIterations=10):
    U = np.zeros(len(mdp.S))
    U_old = copy.copy(U)
    for t in range(numIterations):
        #type your code here
     U_old = U.copy()
    for s in mdp.S:
            U[s] = mdp.R[s, policy] + mdp.discount * np.dot(
                mdp.T[s, policy, :], U_old
            )

    return U

numIterations = 5
pl.figure(figsize=(15,3))
pl.suptitle('Utilities', fontsize=15)
for a in range(4):
    pl.subplot(1,4,a+1)
    U = iterativePolicyEvaluation(mdp=mdp, policy=a, numIterations=numIterations)
    mdp.gridPlot(ax=pl.gca(), im=U.reshape(10,10), title='a='+mdp.A[a], cmap='jet')
pl.show()
#print(np.round(U.reshape(10,10),3))
```

### 2. Policy iteration

Policy evaluation can be used in policy iteration:
1. Given the current policy, compute U
2. Using U, compute a new policy

```python
def policyIteration(mdp, numIterations=1):
    U_pi_k = np.zeros(len(mdp.S)) #initial values
    pi_k = np.random.randint(low=0,high=4,size=len(mdp.S),dtype=int) #initial policy
    pi_kp1 = copy.copy(pi_k)
    for t in range(numIterations):
        #Policy evaluation: compute U_pi_k
          #type your code here
           for _ in range(100):
            U_old = U_pi_k.copy()
            for s in mdp.S:
                a = pi_k[s]
                U_pi_k[s] = mdp.R[s, a] + mdp.discount * np.dot(
                    mdp.T[s, a, :], U_old
                )

            if np.max(np.abs(U_pi_k - U_old)) < 1e-6:
                break
        #Policy improvement
          #type your code here
            for s in mdp.S:
             action_values = [
                mdp.R[s, a]
                + mdp.discount * np.dot(mdp.T[s, a, :], U_pi_k)
                for a in range(len(mdp.A))
            ]
            pi_kp1[s] = np.argmax(action_values)

           if np.array_equal(pi_kp1, pi_k):
            break

            pi_k = pi_kp1.copy()
    return U_pi_k, pi_kp1

U_pi_k, pi_kp1 = policyIteration(mdp, numIterations=2)
```
